```python
import numpy as np
```

```python
emp=np.array(["Arun","Bhavana","Charan","Divya","Eshan","Fathima"])
performance=np.array([[78,82,85,88],
                      [65,70,68,74],
                      [92,89,95,91],
                      [55,62,60,67],
                      [81,79,84,87],
                      [72,75,78,80]
                     ])
quarters=np.array(["Q1","Q2","Q3","Q4"])
print(emp)
print(performance)
print(quarters)
```

```python
total=performance.sum(axis=1)
print(total)
```

```python
avg=performance.mean(axis=1)
print(avg)
```

```python
avg_performance=performance.mean(axis=0)
print(avg_performance)
```

```python
total=performance.sum(axis=1)
score=np.argmax(total)
print(emp[score])
print(total[score])
print(f'{emp[score]} with {total[score]}')
```

```python
total=performance.sum(axis=1)
rank=np.argsort(total)[::-1]
print(rank)

for i in rank:
    print(emp[i],total[i])
```
