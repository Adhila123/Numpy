```python
#3.	What are the total sales, total profit, and average order-line sales?

print("Total_Sales:",sp["Sales"].sum())
print("Total_Profit:",sp["Profit"].sum())
print("Avg_Sales:",sp["Sales"].mean())
```

```python
#4.	Which product categories generate the most sales?

category_sales=sp.groupby("Category")["Sales"].max()
print(category_sales.sort_values(ascending=False))
```

```python
#5.	Which sub-categories generate the highest and lowest profit?

sub_profit=sp.groupby("Sub-Category")["Profit"].sum()
print(sub_profit.sort_values(ascending=False))
```

```python
sub_profit.idxmax()
```

```python
sub_profit.idxmin()
```

```python
#6.	Which region has the highest sales?

region_sales=sp.groupby("Region")["Sales"].sum()
print(region_sales.sort_values(ascending=False))

region_sales.idxmax()
```

```python
#7.	Which customer segment contributes the most sales?

segment_sales=sp.groupby("Segment")["Sales"].sum()
print(segment_sales.sort_values(ascending=False))

segment_sales.idxmax()
```

```python
#8.	How do discounts relate to profit?

disc_profit=sp.groupby("Discount")["Profit"].sum()
print(disc_profit.sort_values(ascending=False))
```
