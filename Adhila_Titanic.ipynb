**Q11. Combining DataFrames:concat,merge and join**

**(a)**

```python
part_a=df.loc[1:4,['Record_ID','class','fare']]

part_b=df.loc[5:8,['Record_ID','class','fare','age']]

result=pd.concat(
    [part_a,part_b],
    ignore_index=True)

result
```

concat() combines both DataFrames vertically and keeps all columns.Missing values are filled with Nan.

**(b)**

```python
result=pd.concat(
    [part_a,part_b],
    join='inner',
    ignore_index=True)

result
```

**(c)**

```python
profile=df.loc[1:6,['Record_ID','class','fare']]
scores=df.loc[4:9,['Record_ID','fare','age']]
merged=pd.merge(
    profile,scores,
    on='Record_ID',
    suffixes=('_profile','_scores'))

merged
```

**(d)**

```python
result=pd.merge(
    profile,scores,
    on='Record_ID',
    how='outer',
    indicator=True)
result
```

**(e)**

```python
left_join=profile.set_index('Record_ID').join(
    scores.set_index('Record_ID'),
    how='left',
    lsuffix='_profile',
    rsuffix='_scores')

left_join
```
