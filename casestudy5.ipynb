```python
#D1
ratings=pd.Series(rating_list)
print(len(ratings))
print(round(ratings.mean(),2))
print(ratings.max())
print(ratings.min())
```

```python
#D2
print(ratings.value_counts().sort_index())
```

```python
#D3
def sentiment(r):
    if r>=4:
        return('Positive')
    elif r==3:
        return('Neutral')
    else:
        return('Negative')
print(ratings.apply(sentiment))
```

```python
#D4
sentiments=(ratings.apply(sentiment))
print(round(sentiments.value_counts(normalize=True)*100,1))
```

```python
#E1
def portfolio_action(pct):
    if pct>=20:
        return('Expand')
    elif pct>=10:
        return('Maintain')
    else:
        return('Revision')
print(share.apply(portfolio_action))
```
