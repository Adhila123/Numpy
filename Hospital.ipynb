```python
# 13.Detect outliers in BillAmount using the IQR method. Show the limits and the outlier rows. Then compute the correlation between StayDays and 
#BillAmount again after removing the outlier. What changed? Which version would you trust?

Q1=df['BillAmount'].quantile(0.25)
Q3=df['BillAmount'].quantile(0.75)

IQR=Q3-Q1

lower= Q1 - 1.5 * IQR
upper= Q3 + 1.5 * IQR

outliers = df[
    (df['BillAmount'] < lower) | (df['BillAmount'] > upper)
    ]

print(outliers)
```

```python
df_clean=df[(df['BillAmount']>=lower)&(df['BillAmount']<=upper)].copy()

print(df_clean['StayDays'].corr(df_clean['BillAmount']))
```

IQR method is used to find and remove unusually high or low bill amounts.
