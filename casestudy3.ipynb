```python
#13. Write a function grade(score) that returns:
#a. 90 and above: "Distinction"
#b. 75 to 89: "First Class"
#c. 60 to 74: "Second Class"
#d. below 60: "Fail"

def grade(s1):
    if s1>=90:
        return('Distinction')
    elif s1>=75 and s1<=89:
        return('First Class')
    elif s1>=60 and s1<=74:
        return('Second Class')
    else:
        return('Fail')
print(s1.apply(grade))
```

```python
#14. Display how many students fall in each grade, then the same as percentages.
print(s1.value_counts())

print(s1.value_counts(normalize=True)*100)
```
