3.   cal revenue (unit sol*unit price)
4.   find the total reveue per product for the quarter
5.   find the month with highest total revenue
6.   save as quarterly_sales.csv

```python
import pandas as pd
```

```python
# 1.load the 3 monthly files
jan=pd.read_csv("sales_jan.csv")
feb=pd.read_csv("sales_feb.csv")
mar=pd.read_csv("sales_mar.csv")
```

```python
# 2.stack them into on dataset and record which month each row came from

jan["Month"]="January"
feb["Month"]="February"
mar["Month"]="March"

df=pd.concat([jan,feb,mar],ignore_index=True)
print(df)
```

```python
# 3.cal revenue (unit sol*unit price)

df["Revenue"]=df["Units_Sold"]*df["Unit_Price"]
print(df)
```

```python
# 4.find the total reveue per product for the quarter

prod_revenue=df.groupby("Product")["Revenue"].sum()
print(prod_revenue)
```

```python
# 5.find the month with highest total revenue

monthly_revenue=df.groupby("Month")["Revenue"].sum().idxmax()
print(monthly_revenue)
```

```python
# 6.save as quarterly_sales.csv

df.to_csv("quarterly_sales.csv",index=False)
```
